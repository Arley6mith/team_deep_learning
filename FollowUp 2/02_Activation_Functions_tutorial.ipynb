### Activation Functions = Transfer Functions = 'Funciones de paso'= De No linealidad

Activation functions are used to introduce non-linearities in the linear output of the type w * x + b in a neural network.
Other way around ANN were simply linear combinations of their parameters.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def graph(y,graph_name):
    x = np.linspace(-10, 10, 1000)
    plt.figure(figsize=(8, 4))
    plt.plot(x, y, color='orange')
    plt.legend([graph_name], shadow=True)
    plt.show()
#type(graph)
```

```python
# lambda functions: la x : que hace la x
# nombre parámetros : cálculos
y = lambda x : max(x,0)
type(y)
```

```python
# Más parecido al calculo sería:
# def f(x) : que hace la x
def ReLU(x) : return np.maximum(x,0)

x = np.linspace(-10, 10, 1000)
graph(ReLU(x), 'Relu')
```

```python
def cuadratic(a, x, b, c, d)  :  return a*x**b + c*x + d  # a the coefficient, b is the exponent and c is the intercept or bias

x = np.linspace(-10, 10, 1000)
graph(cuadratic(1,x,6,100,1000), 'cuadratica')
graph(cuadratic(100,x,2,10,10), 'cuadratica')
plt.plot(x, cuadratic(0.9,x,4,10,100), x, cuadratic(1,x,4,4,40))
plt.show()
```

In a ANN the information is propagated through the network up to the output layer, where the softmax function yields the probability of a given observation belonging to each class.  (Used most commonly in single label classification)

The softmax of y_i is simply the exponential divided by the sum of exponential of the whole Y vector (array):
![image.png](img/softmax.png)
Where S(y_i) is the softmax function of y_i and e is the exponential and j is the no. of columns in the input vector Y.

```python
def softmax(x):
    """Compute softmax values for each
     sets of scores in x."""  # This is called the docstring of a function!
    return np.exp(x) / np.sum(np.exp(x), axis=0)

scores = [0.0, 1.0, -0.85]
print(softmax(scores))

# Depending on threshold, i.e. t = 0.5, the existing classes on the image are?
```
